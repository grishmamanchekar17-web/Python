```python
class hospital():
    def __init__(self,p_name,p_age,disease,blood_group):     # non paramatic constructor
        self.patient=     p_name
        self.age=         p_age
        self.disease=     disease
        self.blood_group= blood_group
    def details(self):     # Method 1
        print(f"patient {self.patient} whose age is {self.age} has come for  the trestment of {self.disease}.")
    def patient_blood_group(self):        # Method 2
        print(f"blood_group of patient {self.patient} is {self.blood_group}")
```

# Inheritance

```python
class Employee():
    def __init__(self,ID,Name,Adhar):    # constructor
        print("Parent constructor has run")
        self.ID     = ID
        self.Name   = Name
        self.id_proof=Adhar

    def WELCOME(self):
        print(f"welcome {self.Name}")

class Department(Employee):
    def __init__(self,ID,Name,Adhar,dept,salary):
        print("Child constructor has  run")
        # super function calls the parent's constructor
        super().__init__(ID,Name,Adhar)    # will read all these attributes
                                     # from parent class
        self.Department=dept
        self.payment=salary
    def info(self):
        print(f"The department of {self.Name} is {self.Department}")
        print(f"The diwali bonus of {self.Name} is {(self.payment*0.05)}")

        # if there is same name eg info in both the class child class will ignore parent class.
        
        
        
```

```python
p1=Employee(1001,"john doe",1234567)
```

```python
C1=Department(1002,"marry smith",7857674733,"HR",50000)
```

```python
C1.WELCOME()
C1.info()
```

```python
p1.info()
```

# Abstraction

```python
from abc import ABC, abstractmethod
```

```python
class company(ABC):   # Abstract- parent class
                      # Abstract method
    @abstractmethod
    def info(self):     # method 1
        print("information")
        
    def details(self):  # method 2
        print("details")

class dept_hr(company):   # child class
    def KPI(self):
        print("KPI")
    def info(self):
        pass

class dept_IT(company):   # child class
    def KPI(self):
        print("KPI")
    def info(self):
        print("hello IT")

IT_001=dept_hr()          # object creation
IT_001.KPI()              # object cannot use the class unless info method is implemented in child  class
```

```python
IT_001=dept_IT()
```

```python
IT_001.info()
```
