```python
try:
    x=int(input("enter dividend"))
    y=int(input("enter divisor"))
    z=x/y
except BaseException as e:
    print("e")

print(z)
```

```python
try:
    x=int(input("enter dividend"))
    y=int(input("enter divisor"))
    z=x/y
except ZeroDivisionError as ZDE:
    print(ZDE)
except ValueError as VE:
    print(VE)

finally:
    print("program ends here thank you")
    

```

```python
try:
    print("enter all codes that can run into error,here:")
except:
    print("except runs whenever error occurs in the try codes,")
else:
    print("else runs only when there is no error in the try codes")
finally:
    print("finally runs always,irrespective of error occurred or not")
```
